# Phase 3 — Evaluation & Analysis

Metrics and baselines for the 29-problem LLM-hard run. The headline numbers are
read from the saved run's metrics (**no API key needed**); the majority-vote
baseline implementation is demonstrated on the deterministic (number /
multiple-choice) subset.

```python
# --- setup: make the `debate` package importable and resolve project paths ---
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))   # fallback if `pip install -e .` wasn't run

DATA = ROOT / "data"
OUTPUTS = ROOT / "outputs"
```

```python
import json
import pandas as pd

run = json.loads((OUTPUTS / "hard_result.json").read_text())
m = run["metrics"]
n = len(run["evaluations"])
print(f"{n} problems  ·  participants:")
for slot, spec in run["config"]["participants"].items():
    print(f"  {slot}: {spec['model']}")
```

## System metrics

- **accuracy** — the debate's final (judge-selected) answer
- **improvement_rate** — refinement turned a wrong answer right
- **consensus_rate** — all three solvers agreed after refinement
- **judge_accuracy_on_disagreements** — when solvers disagree, does the judge pick correctly?

```python
labels = {
    "accuracy": "Debate accuracy",
    "improvement_rate": "Improvement rate",
    "consensus_rate": "Consensus rate",
    "judge_accuracy_on_disagreements": "Judge acc. (disagreements)",
}
pd.DataFrame(
    [(v, m.get(k)) for k, v in labels.items()], columns=["metric", "value"]
).set_index("metric")
```

## Baseline comparison (the key result)

Three ways to answer, from the same models:

- **Single-LLM** — each solver alone, pre-debate (averaged)
- **Majority vote** — plurality of the 3 independent answers (no refinement, no judge)
- **Full debate** — this system

All three are implemented in `src/debate/evaluation.py`.

```python
baselines = pd.DataFrame({
    "approach": ["Single-LLM", "Majority vote", "Full debate"],
    "accuracy": [m["single_llm_accuracy"], m["majority_vote_accuracy"], m["accuracy"]],
}).set_index("approach")
baselines
```

### Required plot — baselines vs the debate system

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 4))
colors = ["#B0B0B0", "#B0B0B0", "#4C72B0"]
bars = ax.bar(baselines.index, baselines["accuracy"], color=colors)
ax.set_ylim(0, 1.0)
ax.set_ylabel("accuracy")
ax.set_title(f"Accuracy by approach — {n} LLM-hard problems")
for b, v in zip(bars, baselines["accuracy"]):
    ax.text(b.get_x() + b.get_width() / 2, v + 0.02, f"{v:.2f}", ha="center")
plt.tight_layout()
plt.show()
```

The debate beats **both** required baselines. Majority vote is even worse than a
single LLM here: on trap problems the solvers often converge on the *same* wrong
answer, so voting reinforces the mistake — the judge + refinement are what
recover the correct answer.

## The majority-vote baseline, demonstrated

```python
from debate.evaluation import calculate_majority_vote_accuracy
from debate.schemas import DebateResult
from debate.problem import Problem

def parse(v):
    return json.loads(v) if isinstance(v, str) else v

def to_result(r):
    r = dict(r)
    for k in ("role_preferences", "solutions", "refined_solutions"):
        r[k] = {kk: parse(vv) for kk, vv in r[k].items()}
    r["judgment"] = parse(r["judgment"])
    return DebateResult(**r)

results = [to_result(r) for r in run["results"]]
problems = {p["id"]: Problem(**p) for p in json.loads((DATA / "llm_hard_problems.json").read_text())}
by_id = {r.problem_id: problems[r.problem_id] for r in results}

# Deterministic subset (number / multiple_choice) needs no grader / API key.
det = [r for r in results if by_id[r.problem_id].answer_type in ("number", "multiple_choice")]
det_by_id = {r.problem_id: by_id[r.problem_id] for r in det}
print("majority-vote accuracy on the", len(det), "number/MC problems:",
      round(calculate_majority_vote_accuracy(det, det_by_id), 3))
print("(full-set value incl. free_answer, from the saved run: %.3f)" % m["majority_vote_accuracy"])
```
