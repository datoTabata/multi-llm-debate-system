# Phase 2 — The Debate System

Four LLMs collaborate: three **Solvers** answer independently, peer-review each
other, refine, and a fourth **Judge** picks the best answer.

**Pipeline** (`src/debate/pipeline.py`): Role selection → Round 1 (independent
solutions) → Peer review → Round 2 (refinement) → Judgment.

This notebook walks one problem through all five stages. The main walkthrough
reads a **saved run** (no API key needed); an optional cell at the end runs a
live debate if `OPEN_ROUTER_KEY` is set.

```python
# --- setup: make the `debate` package importable and resolve project paths ---
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))   # fallback if `pip install -e .` wasn't run

DATA = ROOT / "data"
OUTPUTS = ROOT / "outputs"
```

```python
import json
from debate.models import load_config

config = load_config()
print("Participant pool (config/models.toml):")
for slot, spec in config.participants.items():
    print(f"  {slot}: {spec.model}")
print("Grader:", config.grader.model)
```

## One problem through the pipeline (from a saved run)

```python
run = json.loads((OUTPUTS / "hard_result.json").read_text())

def parse(v):  # saved runs may store nested fields as JSON strings
    return json.loads(v) if isinstance(v, str) else v

# Feature the birthday-paradox trap if present, else the first result.
res = next((r for r in run["results"] if r["problem_id"] == "hard_024"), run["results"][0])
pid = res["problem_id"]
problem = next(p for p in json.loads((DATA / "llm_hard_problems.json").read_text()) if p["id"] == pid)
print(pid, "-", problem["question"])
print("correct answer:", problem["correct_answer"])
```

### Stage 0 — Role selection\nEach model self-assesses Solver vs Judge; highest judge-confidence becomes the Judge.

```python
for slot, pref in res["role_preferences"].items():
    p = parse(pref)
    tag = "  <- JUDGE" if res["roles"]["judge"] == slot else ""
    print(f"{slot}: wants {p['preferred_role']:6s}  (solver {p['solver_confidence']}, judge {p['judge_confidence']}){tag}")
print("\nassigned roles:", res["roles"])
```

### Stage 1 — Independent solutions

```python
for sid, sol in res["solutions"].items():
    print(f"{sid}: {parse(sol)['answer']!r}")
```

### Stage 2 — Peer review (each solution reviewed by the other two)

```python
for sid, reviews in res["peer_reviews"].items():
    print(f"reviews of {sid}: {len(reviews)}")
print("\nexample review (truncated):\n", res["peer_reviews"][list(res["peer_reviews"])[0]][0][:300])
```

### Stage 3 — Refinement

```python
for sid, ref in res["refined_solutions"].items():
    print(f"{sid}: {parse(ref)['refined_answer']!r}")
```

### Stage 4 — Judgment

```python
j = parse(res["judgment"])
print("winner:", j["winner"], "| confidence:", j["confidence"])
print("reasoning:", j["reasoning"][:300])
```

## Optional: run a live debate

Runs one problem through the real pipeline via OpenRouter. Requires
`OPEN_ROUTER_KEY` in `.env`. Uses `nest_asyncio` so `asyncio` works inside
Jupyter.

```python
import os, asyncio
if os.getenv("OPEN_ROUTER_KEY"):
    import nest_asyncio; nest_asyncio.apply()
    from dotenv import load_dotenv; load_dotenv(ROOT / ".env")
    from debate.models import create_model_clients
    from debate.pipeline import run_debate_for_problem
    from debate.problem import Problem

    models = create_model_clients(config)
    prob = Problem(**problem)
    live = asyncio.run(run_debate_for_problem(prob, models))
    print("roles:", live.roles)
    print("judge picked:", live.judgment.winner, "->",
          live.refined_solutions[live.judgment.winner].refined_answer)
else:
    print("OPEN_ROUTER_KEY not set — skipping live debate (saved walkthrough above is complete).")
```
