# Phase 1 — Problem Dataset

The debate system is evaluated on **challenging problems where a single LLM
often fails**. This notebook explores the two datasets, their categories, and
how each answer is graded.

```python
# --- setup: make the `debate` package importable and resolve project paths ---
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))   # fallback if `pip install -e .` wasn't run

DATA = ROOT / "data"
OUTPUTS = ROOT / "outputs"
```

```python
import json
from collections import Counter
import pandas as pd

def load(name):
    return json.loads((DATA / name).read_text())

assignment = load("problems.json")           # the 25-problem assignment set
llm_hard   = load("llm_hard_problems.json")  # "Easy Problems That LLMs Get Wrong" (arXiv 2405.19616)

print(f"problems.json          : {len(assignment)} problems")
print(f"llm_hard_problems.json : {len(llm_hard)} problems")
```

## Answer types

Every problem declares an `answer_type` that decides how an answer is graded
(`Problem.check` in `src/debate/problem.py`):

- **number** — numeric compare (units stripped, fractions handled, rel-tol 1e-3)
- **multiple_choice** — normalized string match
- **free_answer** — graded by an LLM for semantic equivalence

```python
def summarize(problems, title):
    df = pd.DataFrame(problems)
    print(title)
    print("  by answer_type:", dict(Counter(df["answer_type"])))
    print("  categories    :", len(set(df["category"])))
    return df

df_hard = summarize(llm_hard, "llm_hard_problems.json")
df_assign = summarize(assignment, "problems.json")
```

```python
# One example per answer_type from the LLM-hard set
for atype in ["number", "multiple_choice", "free_answer"]:
    ex = next((p for p in llm_hard if p["answer_type"] == atype), None)
    if ex:
        print(f"\n[{atype}]  {ex['id']}")
        print("Q:", ex["question"][:160])
        print("A:", ex["correct_answer"])
        print("why:", ex.get("explanation", "")[:160])
```

## Distribution of answer types (required plot)

```python
import matplotlib.pyplot as plt

counts = Counter(p["answer_type"] for p in llm_hard)
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(list(counts.keys()), list(counts.values()), color="#4C72B0")
ax.set_title(f"llm_hard_problems.json — answer types ({len(llm_hard)} problems)")
ax.set_ylabel("count")
for i, (k, v) in enumerate(counts.items()):
    ax.text(i, v + 0.2, str(v), ha="center")
plt.tight_layout()
plt.show()
```

## Grading is type-aware

`Problem.check` grades deterministically for `number` / `multiple_choice`
(no API call) and delegates `free_answer` to an LLM grader.

```python
from debate.problem import Problem

p = Problem(id="demo", category="math", question="2+2?", correct_answer="4", answer_type="number")
print("check('4')      ->", p.check("4"))
print("check('four')   ->", p.check("four"))   # no grader -> string fallback
print("check('4.0')    ->", p.check("4.0"))    # numeric tolerance
```
